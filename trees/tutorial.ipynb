```python
import quimb as qu
import quimb.tensor as qtn
```

Creating a tensor 


```python
data = qu.bell_state('psi-').reshape(2, 2)
print(data)
inds = ('k0', 'k1')
tags = ('KET',)

ket = qtn.Tensor(data=data, inds=inds, tags=tags)
ket
```

```python
ket.draw()
```

Create a ket 

```python
bra = qtn.Tensor(qu.rand_ket(4).reshape(2, 2), inds=('b0', 'b1'), tags=['BRA'])
print(bra)
```
